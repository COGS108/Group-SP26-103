```python
# Import libraries for data analysis
import seaborn as sns
sns.set(font_scale=2, style="white")

import matplotlib
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.style as style
# set plotting size parameter
plt.rcParams['figure.figsize'] = (17, 7)

# import pandas & numpy library
import pandas as pd
import numpy as np
import scipy.stats as stats

# Statmodels
import statsmodels
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Load in datasets so the above downloads, cleaning, and wrangling does not need to be done again
df1 = pd.read_csv('data/02-processed/dataset1_FINAL.csv')
df2 = pd.read_csv('data/02-processed/dataset2_FINAL.csv')

# Remove rows without answers
df1 = df1.dropna()
df2 = df2.dropna()

# Do a little cleanup to make things easier later. The dataframe stores numeric data as floats, but the values are all integers except for isolation_t_score
df1['Age'] = df1['Age'].astype(int)
df1['mental_distress_score'] = df1['mental_distress_score'].astype(int)
df1['social_media_freq'] = df1['social_media_freq'].astype(int)
df1['rural_urban_code'] = df1['rural_urban_code'].astype(int)

df2['Age'] = df2['Age'].astype(int)
df2['mental_distress_score'] = df2['mental_distress_score'].astype(int)
df2['social_media_freq'] = df2['social_media_freq'].astype(int)
df2['rural_urban_code'] = df2['rural_urban_code'].astype(int)
```

A weaker connection is shown between Mental Distress Score and Rural-Urban Code, showing that there may be a connection between more Rural living respondants and higher reported mental distress.

Important note: The number of respondants scoring high on social media frequency (higher score = LESS social media usage), and the number of respondants reporting that they live in rural areas (high rural-Urban Code) is quite low. This could result in misleading conclusions.

#### Visual Analysis
There appears to be a high correlation between respondants who report a more isolated lifestyle and a higher level of mental distress. However, care must be taken to consider which causes which. Social isolation can lead to a worsened mental state, but it is also true that social isolation is a symptom of many mental disorders. More analysis is needed to determine if one of these causes the other, or if there is a third confounding variable that affects both.

Something interesting happens when we look at mental distress and isolation as a function of Rural-Urban Code. There is a positive correlation between living more rural and feeling more isolated, and also between living more rural and higher mental distress.

### Analytical Exploration
The visual tests are good for a qualitative exploration, but it isn't the whole picture. Here we will do a quantitative analysis.

```python
#Transform right skewed data with log transformations to help normalize it for ANOVA and OLS
df_combined['social_media_freq_log'] = np.log1p(df_combined['social_media_freq'])
df_combined['rural_urban_code_log'] = np.log1p(df_combined['rural_urban_code'])
```

```python
# OLS tests
test1 = smf.ols('mental_distress_score ~ social_media_freq', data = df_combined).fit()
test1_log = smf.ols('mental_distress_score ~ social_media_freq_log', data = df_combined).fit()

test2 = smf.ols('mental_distress_score ~ rural_urban_code', data = df_combined).fit()
test2_log = smf.ols('mental_distress_score ~ social_media_freq_log', data = df_combined).fit()

test3 = smf.ols('mental_distress_score ~ social_media_freq + rural_urban_code', data = df_combined).fit()
test3_log = smf.ols('mental_distress_score ~ social_media_freq_log + rural_urban_code_log', data = df_combined).fit()

print('Mental Distress Score as a result of Social Media Frequency: ')
print(test1.summary())
print(test1_log.summary())
print('\n\nMental Distress Score as a result of Rural-Urban Code: ')
print(test2.summary())
print(test2_log.summary())
print('\n\nMental Distress Score as a combined result of Social Media Frequency and Rural-Urban Code: ')
print(test3.summary())
print(test3_log.summary())
```

### Results:
Out of all the tested variables, the only one that shows any sort of reasonable statistical significance is Rural-Urban Code. Out of all the Least Squares models evaluated, it had the highest Adjusted R-Squared value, which was 0.138. 
It is also the only variable that has a Prob(>F) that is LESS than 0.05. The Prob(>F) score is 0.0380
