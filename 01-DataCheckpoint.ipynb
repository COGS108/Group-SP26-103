## Background and Prior Work

The relationship between social media use and mental health has rapidly become one of the most heavily examined topics in psychological research over the past decade. As of now, more than 30% of college students report experiencing depression, and approximately 90% of young adults today report utilizing social media in their daily lives – roughly eight times the proportion that did so in 2005. Due to the speed of which this percentage changed, researchers across the globe have dedicated tremendous effort to further examining whether social media use contributes to rising depression and anxiety levels among college-aged students.

Young Adult Depression and Anxiety Linked to Social Media Use: Assessment and Treatment” has examined the relationship between social media use and mental health specifically among college-aged adults. Bettmann reviewed existing literature on this topic and found that social media use is consistently associated with depression and anxiety in young adults.1 The authors showed that as many as 90% of young adults currently use social media compared to only 12.5% in 2005. It also highlighted the fact that more than 30% of college students are currently depressed.1 Significant results of the research were that it is not simply the amount of time spent on social media that matters. However, how you use really mattered. Aspects such as passive scrolling and frequent social comparison were particularly strongly linked to depressive and anxious symptoms. Moreover, young adults who used a larger variety of platforms were significantly more likely to show symptoms of depression and anxiety compared to those who used only few. 1 The authors also introduced the concept of Fear of Missing Out (FoMO), which is the anxiety that others are having more rewarding experiences, as a key mechanism that links social media usage to depression and anxiety. However, the review also acknowledged that the direction of cause and effect remains unclear. It is questionable whether social media use leads to worse mental health, or whether individuals who already experience depression and anxiety are simply more drawn to heavy social media use. This ambiguity would also be an important factor that our project should consider.

In an attempt to empirically examine how social media use correlates with mental health indicators, Osman conducted a cross-sectional study at Mogadishu University. Out of the 268 participants he interviewed, 84.7% reported spending more than three hours per day using social media.2 High levels of social media usage proved a strong correlation between sleep disturbance, anxiety, and social isolation. Of the three, anxiety outcomes proved to be the highest (AOR: 22.2; 95% CI: 3.73-131.90).2 These results are particularly noteworthy, as the finding regarding anxiety outcomes showcases that independent screen usage – despite specific content viewed or platform type – may be enough information to predict anxiety levels in a population made up of college-aged students.

On a grander scale, Wang, Zhang, and Xiao (2025) further investigated these associations in a large-scale cross-cultural study of 1,500 Chinese university students drawn from ten universities across China.3 Utilizing the Depression Anxiety Stress Scale-21 (DASS-21) and the Social Media Use Integration Scale (SMUIS), the authors found that there were indeed positive associations between social media usage and both depressive and anxiety symptoms. This study also identified sleep quality as a significant mediator within this relationship, as disrupted sleep proved to account for roughly 30.4% of the total effect of social media use on depression. This suggests that disrupted sleep could be a manner through which heavy platform use actively degrades mental health.3 Additionally, the association between social media usage and the presence of anxiety and depressive symptoms was found to be higher among female students, a finding that should be treated as a potential confounding variable in further research/analysis done on this topic.3

In totality, each of these studies suggest that there is a consistent correlation between social media use and depression and anxiety symptoms in college-aged individuals. These studies also go out of their way to acknowledge important moderating factors, such as gender and sleep quality, and are valid by utilizing validated screening instruments and rigorous sampling methods. Building upon this prior work, our project will further investigate whether living environment (urban versus rural) influences the association between social media usage and mental health outcomes among young adults in the United States.

By incorporating urbanicity as an additional factor, we aim to provide a more nuanced analysis of how social media usage may relate differently to anxiety and depression symptoms across populations.

### Dataset #1: HINTS 7 — Health Information National Trends Survey (2024)

> Dataset Descriptions: 

This dataset is from the 2024 Health Information National Trends Survey (HINTS 7). It is a nationally representative study conducted by the National Cancer Institute. The purpose of our group using this data set is to explore the association between the social media usage and mental health for college aged(18-25) young adults living in geographical rural areas vs urban areas. Since this is a recent 2024 release, it provides us with a very modern look at these trends.

The dataset tracks several critical metrics that quantify the necessary variables. 

The primary response variable is the phq4 score, which is a composite mental distress metric ranging from 0 to 12. This score combines two items for depression(2 items of PHQ-2) and two for anxiety(2 items from GAD-2). A score of 0–2 is generally considered normal, 3–5 indicates mild distress, and any score of 6 or higher suggests a need for a professional clinical evaluation. Scores of 9 or above are considered severe. 

We also look at the PROMIS_Isolation_t score, which is a standardized T-score for social isolation. It is a standardized metric derived from a specific "PROMIS" (Patient-Reported Outcomes Measurement Information System) scale. It is calculated based on how often a person feels the following four things, which are 'Left out.', 'That people barely know them.', 'Isolated from others.', and 'That people are not "with" them.' Those scores are are scaled relative to the general U.S. population. A score of 50 is exactly the national average.
<br><br>

For our independent variables, we use SocMed_Visited, which is the answer of the question "In the last 12 months, how often did you visit a social media site?" in the scale of 1 to 5. Each scale is 

- 1: Almost every day.

- 2: At least once a week.

- 3: A few times a month.

- 4: Less than once a month.

- 5: Never.
<br><br>

For our third variable(rural vs urban) we used the RUC2023 code, which classifies a respondent's home on a 9-point scale.
Each scale is

- Values 1–3: Represent different sizes of metropolitan (urban) areas.

- Values 4–9: Represent non-metropolitan (rural) areas, ranging from small towns adjacent to cities to completely isolated rural land.

While this dataset is powerful, we have a few significant concerns about using it for this specific project. First, the whole dataset has over 7,000 respondents, our target age group (18–25) is a small subsample of only about 354 respondents. Our concern is the limited statistical power of the size of the subset. Second, everything here is self-reported. This means there are possibilities that some respondents might under-report(or over-report) mental health struggles. Also, there can be respondents that don't unconsciously check social media, under-reporting the social media questionares. Also, a difference between frequency and duration is distinct. The survey asks how often someone visits a site, but not how many hours they actually spend on it, which are two different things.
<br><br>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os


# ============================================================
# STEP1: Loading Data
# ============================================================
DATA_PATH = 'data/00-raw/hints7_public.dta'

assert os.path.exists(DATA_PATH), (
    f"Data file not found at {DATA_PATH}\n"
    "Follow the instructions above to manually download the raw data"
)

# Load the data. Used convert_categoricals=False to keeps numeric codes intact
# This is easier to convert them to NaN later on rather than human-readable categories
df_raw = pd.read_stata(DATA_PATH, convert_categoricals=False)

print(f"Raw dataset loaded successfully.")
# Checking sizes
print(f"Number of observations (rows): {df_raw.shape[0]:,} repspondents")
print(f"Number of variables (columns): {df_raw.shape[1]:,} variables")
```

```python
# ============================================================
# STEP2: Cleaning up data by selecting only the variables relevant to our analysis
# ============================================================

relevant_columns = [
    'Age',

    # --- Dependent variables: mental health ---
    'phq4',
    'PROMIS_Isolation_t',

    # --- Independent variable: social media usage ---
    'SocMed_Visited',

    # --- Third variable: rural/urban (moderator) ---
    'RUC2023',
]

df_subset = df_raw[relevant_columns].copy()
print(f"After column selection: {df_subset.shape[0]:,} respondents × {df_subset.shape[1]} variables")
```

```python
# ============================================================
# STEP 3: Filter to our target population (ages 18-25)
# ============================================================
df_young = df_subset[(df_subset['Age'] >= 18) & (df_subset['Age'] <= 25)].copy()
print(f"After age filter (18-25): {df_young.shape[0]:,} respondents × {df_young.shape[1]} variables")
```

**Missing values analysis**

HINTS encodes missingness using negative integer codes rather than `NaN`.
We've analyzed what each codes mean. 
We will be converting them in 'NaN' to make it look 

| Code | Meaning |
|---|---|
| −9 | Missing data (Not Ascertained) |
| −7 | Missing data (Web partial — Question Never Seen) |
| −5 | Multiple Responses Selected in Error |
| −4 | Unreadable or Non-conforming numeric response |

```python
# ============================================================
# STEP4: Recode HINTS missing-value codes to NaN
# ============================================================
numeric_cols = df_young.select_dtypes(include=[np.number]).columns
n_negatives_before = (df_young[numeric_cols] < 0).sum().sum()
print(f"Number of cells with negative missing code: {n_negatives_before}")

# Recode: replace negatives with NaN
df_young[numeric_cols] = df_young[numeric_cols].where(df_young[numeric_cols] >= 0)

n_negatives_after = (df_young[numeric_cols] < 0).sum().sum()
print(f"Number of negative cells remaining: {n_negatives_after}")
print(f"All HINTS missing codes successfully converted to NaN.")
```

```python
# ============================================================
# STEP 6: Outliers and suspicious values
# ============================================================
print("We will be using Z-score method to find statistical outliers (Z-score > 3)")
variables = ['Age', 'phq4', 'PROMIS_Isolation_t', 'SocMed_Visited','RUC2023'] 

for var in variables:
    mean = df_young[var].mean()
    std_dev = df_young[var].std()
    threshold = 3 * std_dev

    # Calculate Z-scores
    outliers = df_young[(df_young[var] - mean).abs() > threshold]

    print(f"{var}:  Mean: {mean:.2f}  Std: {std_dev:.2f}  Outliers: {len(outliers)}")

df_young.to_csv(
    'data/01-interim/dataset1_interim.csv', index=False
)
```

```python
# Before finalizing the dataset, we will rename the columns to improve readability

rename_cols = {
    'phq4': 'mental_distress_score',
    'PROMIS_Isolation_t': 'isolation_t_score',
    'SocMed_Visited': 'social_media_freq',
    'RUC2023': 'rural_urban_code'
}
df_final = df_young.rename(columns=rename_cols)

df_final.to_csv(
    'data/02-processed/dataset1_FINAL.csv', index=False
)

print("Data Cleaning. Final dataset saved to data/02-processed/dataset1_FINAL.csv")
print("")
print("=====Final dataset preview:=====")
print(f"Number of observations (rows): {df_final.shape[0]:,} repspondents")
print(f"Number of variables (columns): {df_final.shape[1]:,} variables")
print(df_final.head(10))
```
