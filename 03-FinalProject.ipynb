## Abstract

This project looks at whether social media use frequency is associated with anxiety and depression symptoms among young adults ages 18-25 in the United States. We also examine whether this association differs between people living in rural versus urban areas. To answer this question, we use HINTS 6 (2022) and HINTS 7 (2024), two public Health Information National Trends Survey datasets.

We decided to explore this question because these days young adults spend most of their time doing social media, and we wanted to find out how this habit affects their mental health. After receiving a feedback that this initial question is too simple, we've added a third factor of rural versus urban areas to find out if other factor also affects the associated anxiety and depression.

Overall, the results do not show strong support for our original hypothesis. Social media frequency was only weakly related to mental distress, and the rural/urban interaction was not strong enough to make a confident conclusion. The clearest pattern we found was that respondents with higher social isolation scores tended to have higher mental distress scores. However, since this is cross-sectional survey data, it is hard to say that one causes the other.

## Background and Prior Work

The relationship between social media use and mental health has rapidly become one of the most heavily examined topics in psychological research over the past decade. As of now, more than 30% of college students report experiencing depression, and approximately 90% of young adults today report utilizing social media in their daily lives – roughly eight times the proportion that did so in 2005. Due to the speed of which this percentage changed, researchers across the globe have dedicated tremendous effort to further examining whether social media use contributes to rising depression and anxiety levels among college-aged students.

Young Adult Depression and Anxiety Linked to Social Media Use: Assessment and Treatment” has examined the relationship between social media use and mental health specifically among college-aged adults. Bettmann reviewed existing literature on this topic and found that social media use is consistently associated with depression and anxiety in young adults.1 The authors showed that as many as 90% of young adults currently use social media compared to only 12.5% in 2005. It also highlighted the fact that more than 30% of college students are currently depressed.1 Significant results of the research were that it is not simply the amount of time spent on social media that matters. However, how you use really mattered. Aspects such as passive scrolling and frequent social comparison were particularly strongly linked to depressive and anxious symptoms. Moreover, young adults who used a larger variety of platforms were significantly more likely to show symptoms of depression and anxiety compared to those who used only few. 1 The authors also introduced the concept of Fear of Missing Out (FoMO), which is the anxiety that others are having more rewarding experiences, as a key mechanism that links social media usage to depression and anxiety. However, the review also acknowledged that the direction of cause and effect remains unclear. It is questionable whether social media use leads to worse mental health, or whether individuals who already experience depression and anxiety are simply more drawn to heavy social media use. This ambiguity would also be an important factor that our project should consider.

In an attempt to empirically examine how social media use correlates with mental health indicators, Osman conducted a cross-sectional study at Mogadishu University. Out of the 268 participants he interviewed, 84.7% reported spending more than three hours per day using social media.2 High levels of social media usage proved a strong correlation between sleep disturbance, anxiety, and social isolation. Of the three, anxiety outcomes proved to be the highest (AOR: 22.2; 95% CI: 3.73-131.90).2 These results are particularly noteworthy, as the finding regarding anxiety outcomes showcases that independent screen usage – despite specific content viewed or platform type – may be enough information to predict anxiety levels in a population made up of college-aged students.

On a grander scale, Wang, Zhang, and Xiao (2025) further investigated these associations in a large-scale cross-cultural study of 1,500 Chinese university students drawn from ten universities across China.3 Utilizing the Depression Anxiety Stress Scale-21 (DASS-21) and the Social Media Use Integration Scale (SMUIS), the authors found that there were indeed positive associations between social media usage and both depressive and anxiety symptoms. This study also identified sleep quality as a significant mediator within this relationship, as disrupted sleep proved to account for roughly 30.4% of the total effect of social media use on depression. This suggests that disrupted sleep could be a manner through which heavy platform use actively degrades mental health.3 Additionally, the association between social media usage and the presence of anxiety and depressive symptoms was found to be higher among female students, a finding that should be treated as a potential confounding variable in further research/analysis done on this topic.3

In totality, each of these studies suggest that there is a consistent correlation between social media use and depression and anxiety symptoms in college-aged individuals. These studies also go out of their way to acknowledge important moderating factors, such as gender and sleep quality, and are valid by utilizing validated screening instruments and rigorous sampling methods. Building upon this prior work, our project will further investigate whether living environment (urban versus rural) influences the association between social media usage and mental health outcomes among young adults in the United States.

By incorporating urbanicity as an additional factor, we aim to provide a more nuanced analysis of how social media usage may relate differently to anxiety and depression symptoms across populations.

^ Osman, W. A. (2025). Social media use and associated mental health indicators among university students: A cross-sectional study. Scientific Reports, 15, 9534. https://doi.org/10.1038/s41598-025-94355-w

^ Wang, Q., Zhang, J., & Xiao, M. (2025). Association between social media use, physical activity level, and depression and anxiety among college students: A cross-cultural comparative study. Scientific Reports, 15, 6596. https://doi.org/10.1038/s41598-025-88140-y

# Hypothesis


We hypothesize that increased daily social media usage is associated with higher levels of anxiety and depression among young adults in the United States. Specifically, individuals who use social media more frequently (visiting social media sites multiple times per day) are expected to report higher levels of anxiety and depression symptoms.

We also hypothesize that this relationship may be stronger among individuals living in rural areas compared to those living in urban areas. One possible reason is that people living in rural environments may rely more heavily on online social interaction and may have less access to offline social communities and mental health resources. As a result, increased social media usage may have a stronger association with anxiety and depression symptoms among rural populations.

This will be used as our third variable to compare the association between social media and mental health between rural and urban areas.

  - Descriptions of any shortcomings this dataset has with repsect to the project

### Data Cleaning

The full data cleaning process is shown in `01-DataCheckpoint.ipynb`, where each raw HINTS file was loaded, filtered to ages 18-25, missing-value codes were converted to `NaN`, and variables were renamed into a shared format.

For the final report, to prevent the length of be overly long, we are not repeating every single cell from the checkpoint. From here, we load the processed datasets created during the Data Checkpoint and use them for the final EDA and analysis.


```python
# Import libraries for data analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.formula.api as smf

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df1 = pd.read_csv('data/02-processed/dataset1_FINAL.csv')
df2 = pd.read_csv('data/02-processed/dataset2_FINAL.csv')

df1['survey_year'] = 2024
df2['survey_year'] = 2022

df1 = df1.dropna()
df2 = df2.dropna()

df1['Age'] = df1['Age'].astype(int)
df1['mental_distress_score'] = df1['mental_distress_score'].astype(int)
df1['social_media_freq'] = df1['social_media_freq'].astype(int)
df1['rural_urban_code'] = df1['rural_urban_code'].astype(int)

df2['Age'] = df2['Age'].astype(int)
df2['mental_distress_score'] = df2['mental_distress_score'].astype(int)
df2['social_media_freq'] = df2['social_media_freq'].astype(int)
df2['rural_urban_code'] = df2['rural_urban_code'].astype(int)

df_combined = pd.concat([df1, df2], ignore_index=True)
df_combined['rural_urban_group'] = np.where(
    df_combined['rural_urban_code'] <= 3,
    'Urban',
    'Rural'
)
df_combined['rural_binary'] = np.where(df_combined['rural_urban_group'] == 'Rural', 1, 0)

print(f"HINTS 7 complete cases: {len(df1)}")
print(f"HINTS 6 complete cases: {len(df2)}")
print(f"Combined final analysis sample: {len(df_combined)}")
```

```python
# Quick sample-size check
pd.crosstab(df_combined['survey_year'], df_combined['rural_urban_group'])
```

The final analysis uses 603 complete cases after combining HINTS 6 and HINTS 7. One important limitation is that the rural group is much smaller than the urban group, so rural/urban comparisons should be interpreted carefully.

### Rural/Urban Comparison

This plot is newly added for the final report because it directly matches the final research question about whether the social media/mental distress relationship differs by rural versus urban group.


```python
mean_by_group = (
    df_combined
    .groupby(['social_media_freq', 'rural_urban_group'])
    .agg(
        mean_mental_distress=('mental_distress_score', 'mean'),
        n=('mental_distress_score', 'size')
    )
    .reset_index()
)

fig, ax = plt.subplots(figsize=(9, 6))
sns.lineplot(
    data=mean_by_group,
    x='social_media_freq',
    y='mean_mental_distress',
    hue='rural_urban_group',
    marker='o',
    ax=ax
)
ax.set_title('Average Mental Distress by Social Media Frequency and Rural/Urban Group')
ax.set_xlabel('Social Media Frequency (1 = almost every day, 5 = never)')
ax.set_ylabel('Average Mental Distress Score')

plt.tight_layout()
plt.show()

mean_by_group
```

This plot explicitly addresses our research question. If rural/urban status strongly changed the social media and mental health relationship, we would expect the rural and urban lines to show clearly different patterns. However, as shown in the plot, the pattern is not very stable.


### Social Isolation and Mental Distress


```python
fig, ax = plt.subplots(figsize=(9, 6))

sns.regplot(
    data=df_combined,
    x='isolation_t_score',
    y='mental_distress_score',
    scatter_kws={'alpha': 0.35},
    ax=ax
)
ax.set_title('Social Isolation and Mental Distress')
ax.set_xlabel('Isolation T-Score')
ax.set_ylabel('Mental Distress Score')

plt.tight_layout()
plt.show()

corr, p_value = stats.pearsonr(df_combined['isolation_t_score'], df_combined['mental_distress_score'])
print(f"Correlation between isolation and mental distress: r = {corr:.3f}, p = {p_value:.4f}")
```

The strongest pattern in the final EDA is the positive relationship between social isolation and mental distress. Respondents with higher isolation scores also tended to report higher mental distress scores. However, it is hard to say this has a strong correlation because social isolation could contribute to distress, and in opposite direction also(distress could contribute to isolation). Additionally, another factor could affect both(confouding variable).

## Statistical Analysis

The visual analysis is helpful, but we also include a simple regression analysis to summarize whether social media frequency and rural-urban context are associated with mental distress.

To maintain continuity with the EDA checkpoint, we use a group-level regression approach. Each row in the regression dataset represents one observed combination of social media visit frequency and rural-urban continuum code, following the simplified modeling structure used in the EDA checkpoint.

These models should be interpreted as exploratory summaries of the patterns. We also add an interaction model for the final report to directly address our final research question which is whether the relationship between social media frequency and mental distress changes across the rural-urban continuum.

```python
df_checkpoint_regression = df_combined.drop_duplicates(
    ['social_media_freq', 'rural_urban_code']
).copy()


def get_model_stats(data, formula, label):
    model = smf.ols(formula, data=data).fit()
    return {
        'model': label,
        'r_squared': model.rsquared,
        'adj_r_squared': model.rsquared_adj,
        'model_p_value': model.f_pvalue
    }


model_rows = [
    get_model_stats(
        df_checkpoint_regression,
        'mental_distress_score ~ social_media_freq',
        'social media frequency only'
    ),
    get_model_stats(
        df_checkpoint_regression,
        'mental_distress_score ~ rural_urban_code',
        'rural-urban code only'
    ),
    get_model_stats(
        df_checkpoint_regression,
        'mental_distress_score ~ social_media_freq + rural_urban_code',
        'social media + rural-urban code'
    ),
    get_model_stats(
        df_checkpoint_regression,
        'mental_distress_score ~ social_media_freq * rural_urban_code',
        'interaction: social media x rural-urban code'
    ),
]

model_comparison = pd.DataFrame(model_rows)
model_comparison.round(4)
```

Overall, the models should be interpreted cautiously as exploratory summaries rather than causal evidence. The interaction model helps check whether the rural/urban context affects the social media/mental distress pattern, but the results do not provide strong evidence of a clear interaction.

## Discussion and Conclusion

This project examined whether social media use frequency is associated with anxiety and depression symptoms among young adults ages 18-25 in the United States, and whether this association differs between rural and urban respondents. We used HINTS 6 (2022) and HINTS 7 (2024), filtered both datasets to our target age range, cleaned HINTS missing-value codes in the Data Checkpoint, and combined the two processed datasets for the final exploratory analysis.

The results did not show a strong or consistent relationship between social media visit frequency and mental distress. An critical reason was that the most respondents reported using social media almost every day. This led to limited variation in the main predictor. Rural and urban respondents also had similar average mental distress patterns, although the rural group was much smaller. The clearest pattern in the final EDA was that higher social isolation scores were associated with higher mental distress scores.

In `02-EDACheckpoint.ipynb`, it suggested among the possible regression models, the rural-urban code was the most reasonable simple regression predictor.
However, overall, the regression results do not strongly support our hypothesis. Social media frequency was not a strong predictor of mental distress, and the interaction between social media frequency and rural/urban status was not strong enough to conclude that the relationship differs across these two groups. Therefore, our final conclusion is that this dataset does not provide strong evidence for our original expectation that frequent social media use is clearly linked to higher anxiety and depression symptoms, or that the link is clearly stronger for rural young adults.

There are several important limitations. First, the data are cross-sectional. This means it is vague to determine whether social media use affects mental health, whether mental health affects social media use, or whether another factor affects both. Second, the social media variable measures visit frequency, not time spent on social media or type of use. This also makes the social media usage variable vague, the more frequent you visit the social media doesn't necessarily mean you spend most of your time using social media. Third, the rural sample is small, which limits our rural vs urban comparsion.

Future work would be better if we can get data from a larger sample of young adults, especially more rural respondents, and more detailed social media measures such as platform type, and time spent. A longitudinal dataset would also be more useful for understanding whether changes in social media behavior happen before or after changes in mental health. Overall, our project suggests that the relationship between social media and mental health is more complicated. It is hard to simply state that "more social media equals worse mental health".
